# Data Splitting

```python
# takes the 'Annual Income (k$)' and 'Spending Score (1-100)' columns from the dataset and stores them in an array X
X = df.iloc[:, [3, 4]].values
 
# displays the retrieved data in DataFrame format with appropriate column names.
print(pd.DataFrame(X, columns=['Annual Income (k$)', 'Spending Score (1-100)']))
```

# Elbow Method

```python
# initialize the KMeans model without initial parameters
kmeans = KMeans()
 
# initialize the KElbow visualizer to determine the optimal number of clusters.
visualizer = KElbowVisualizer(kmeans, k=(1, 10))
 
# fit visualizer to data to find the optimal number of clusters
visualizer.fit(X)
 
# displays elbow graph for analysis
visualizer.show()
```

# Cluster Modeling (K-Means Clustering)

```python
from sklearn.cluster import KMeans
 
# initialize and train the KMeans model with the number of clusters = 4
kmeans = KMeans(n_clusters=4, random_state=0)
kmeans.fit(X)
 
# get cluster labels
labels = kmeans.labels_
 
# get total custer
k = 4
 
# func for cluster characteristic analysis
def analyze_clusters(X, labels, k):
    print("Analysis of Characteristics of Each Cluster:")
    for cluster_id in range(k):
        # fetch data for current cluster
        cluster_data = X[labels == cluster_id]
 
        # calculate the average for each feature in the cluster
        mean_income = cluster_data[:, 0].mean()  # avarage annual income
        mean_spending = cluster_data[:, 1].mean()  # avarage spending Score 
 
        print(f"\nCluster {cluster_id + 1}:")
        print(f"Avarage Annual Income (k$): {mean_income:.2f}")
        print(f"Avarage Spending Score (1-100): {mean_spending:.2f}")
 
# analysis of the characteristics of each cluster
analyze_clusters(X, labels, k)
```

```python
import matplotlib.pyplot as plt
 
# determine the position of the centroid
centroids = kmeans.cluster_centers_
 
# cluster visualization
plt.figure(figsize=(12, 8))
 
# plot data
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis', s=50, alpha=0.6, edgecolors='w', marker='o')
 
# plot centroid
plt.scatter(centroids[:, 0], centroids[:, 1], c='red', s=200, marker='X', label='Centroids')
 
# add centroid labels to the plot
for i, centroid in enumerate(centroids):
    plt.text(centroid[0], centroid[1], f'Centroid {i+1}', color='red', fontsize=12, ha='center', va='center')
 
# add title and label
plt.title('Cluster Visualization with Centroids')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.legend()
 
plt.show()
 
# display controid values
print("Centroids Value:")
for i, centroid in enumerate(centroids):
    print(f"Centroid {i+1}: Annual Income = {centroid[0]:.2f}, Spending Score = {centroid[1]:.2f}")
```
